```python
import freesasa
import os
import csv
from collections import defaultdict

pdb_folder = "E:\RNA_binding_desktop\dataset_design\Hariboss_TR60"
output_csv = "Hariboss_TR60_sasa_seqindex.csv"

all_rows = []

for pdb_file in sorted(os.listdir(pdb_folder)):
    if not pdb_file.endswith(".pdb"):
        continue

    pdb_path = os.path.join(pdb_folder, pdb_file)
    print("Processing:", pdb_file)

    structure = freesasa.Structure(pdb_path)
    result = freesasa.calc(structure)

    current_chain = None
    current_resnum = None
    current_resname = None
    current_atoms = []

    seq_index_per_chain = defaultdict(int)

    def flush_residue(chain, atoms, resname):
        """Write one residue SASA entry"""
        if not atoms:
            return
        sasa = sum(result.atomArea(i) for i in atoms)
        pos = seq_index_per_chain[chain]
        seq_index_per_chain[chain] += 1

        all_rows.append([
            pdb_file.replace(".pdb", ""),
            chain,
            pos,                 # ✅ sequence index (matches JSON)
            resname,
            round(sasa, 3)
        ])

    for i in range(structure.nAtoms()):
        chain = structure.chainLabel(i)
        resnum = structure.residueNumber(i)
        resname = structure.residueName(i).strip()

        # start of new residue
        if (
            chain != current_chain or
            resnum != current_resnum or
            resname != current_resname
        ):
            flush_residue(current_chain, current_atoms, current_resname)
            current_chain = chain
            current_resnum = resnum
            current_resname = resname
            current_atoms = []

        current_atoms.append(i)

    # flush last residue
    flush_residue(current_chain, current_atoms, current_resname)

# write CSV
with open(output_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["pdb", "chain", "pos", "resname", "sasa"])
    writer.writerows(all_rows)

print("✅ Sequence-indexed SASA written to:", output_csv)
```
